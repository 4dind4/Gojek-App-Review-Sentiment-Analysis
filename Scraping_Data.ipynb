```python
import pandas as pd
import numpy as np
import re
import matplotlib.pyplot as plt
import seaborn as sns
from google_play_scraper import reviews, Sort
```

```python
print("Memulai scraping ulasan...")

result, _ = reviews(
    'com.gojek.app',         # App ID Gojek di Play Store
    lang='id',               # Bahasa Indonesia
    country='id',
    sort=Sort.NEWEST,
    count=6000
)

df_raw = pd.DataFrame(result)
print(f"Total ulasan berhasil di-scrape: {len(df_raw)}")
df_raw.head()
```

```python
df_raw.to_csv('gojek_reviews.csv', index=False)
print('Data berhasil disimpan ke gojek_reviews.csv')
```
